```python
import pandas as pd
import altair as alt
alt.data_transformers.enable('vegafusion')
from geopy.geocoders import ArcGIS
```

```python
df = pd.read_csv("https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/licenses_fall2022.csv")
df.head()
```

```python
df['License Type'].unique()
```

```python
df['License Type'].value_counts()
```

```python
df['Location'] = df['City'].str.cat(df['State'], sep=', ')
df.head()
```

```python
df_nt = df[df["Description"] == 'LICENSED NAIL TECHNICIAN']
```

```python
df_nt = df_nt[df_nt.State == "IL"]
```

```python
ag = ArcGIS()
```

```python
def geocoding_lat(location):
    return ag.geocode(location).latitude
```

```python
def geocoding_long(location):
    return ag.geocode(location).longitude
```

```python
unique_locations = df_nt['Location'].unique()
d = dict(zip(unique_locations, pd.Series(unique_locations).apply(lambda x : geocoding_lat(x))))
df_nt['latitude'] = df_nt['Location'].map(d) 
```

```python
d = dict(zip(unique_locations, pd.Series(unique_locations).apply(lambda x : geocoding_long(x))))
df_nt['longitude'] = df_nt['Location'].map(d)
```

```python
df_nt.head() 
```

```python
df_nt['License Status'].value_counts()
```

```python
df_nt.shape
```

```python
illinois_lines = alt.topo_feature("https://services2.arcgis.com/aIrBD8yn1TDTEXoz/arcgis/rest/services/Counties/FeatureServer/0/query?outFields=*&where=1%3D1&f=geojson", 
                            feature='geometry')
brush = alt.selection_interval()

background = alt.Chart(illinois_lines).mark_geoshape(
    fill='lightgray',
    stroke='white'
).properties(
    width=300,
    height=500
).project('albersUsa')

points = alt.Chart(df_nt[['latitude', 'longitude', 'License Status']]).mark_circle(
    size=30
).encode(
    longitude='longitude:Q',
    latitude='latitude:Q',
    color=alt.when(brush).then("License Status:N").otherwise(alt.value("black"))
).add_params(
    brush
)

bars = alt.Chart(df_nt[['latitude', 'longitude', 'License Status']]).mark_bar().encode(
    y='License Status:N',
    x='count(License Status):Q',
    color='License Status:N'
).transform_filter(
    brush
)

chart = background + points & bars
chart
```
